## DDL Statements, run once per environment

```sql
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog_name || '.ict_foundation');
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog_name || '.ict_metadata');
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog_name || '.ict_audit');
```

```sql
DECLARE OR REPLACE grant_stmt STRING;


SET VAR grant_stmt = 'GRANT CREATE TABLE ON SCHEMA ' || :catalog_name || '.ict_metadata TO `' || :group_name || '`';
EXECUTE IMMEDIATE grant_stmt;
SET VAR grant_stmt = 'GRANT CREATE TABLE ON SCHEMA ' || :catalog_name || '.ict_audit TO `' || :group_name || '`';
EXECUTE IMMEDIATE grant_stmt;
SET VAR grant_stmt = 'GRANT CREATE TABLE ON SCHEMA ' || :catalog_name || '.ict_foundation TO `' || :group_name || '`';
EXECUTE IMMEDIATE grant_stmt;
SET VAR grant_stmt = 'GRANT USE SCHEMA ON SCHEMA ' || :catalog_name || '.ict_metadata TO `' || :group_name || '`';
EXECUTE IMMEDIATE grant_stmt;
SET VAR grant_stmt = 'GRANT USE SCHEMA ON SCHEMA ' || :catalog_name || '.ict_audit TO `' || :group_name || '`';
EXECUTE IMMEDIATE grant_stmt;
SET VAR grant_stmt = 'GRANT USE SCHEMA ON SCHEMA ' || :catalog_name || '.ict_foundation TO `' || :group_name || '`';
EXECUTE IMMEDIATE grant_stmt;

```

```sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog_name || '.ict_metadata.etl_setup_config')
(
    entity_id STRING NOT NULL,
    target_layer STRING,
    source_read_from STRING,
    target_table STRING,
    load_type STRING NOT NULL,
    primary_key_cols ARRAY<STRING>,
    scd_type STRING,
    partition_column STRING,
    raw_retention_days INT,
    load_sequence INT NOT NULL,
    is_active BOOLEAN NOT NULL,
    created_ts TIMESTAMP NOT NULL,
    created_by STRING NOT NULL,
    updated_ts TIMESTAMP,
    updated_by STRING
)
```

```sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog_name || '.ict_metadata.column_mapping')
(   table_name STRING,
    target_layer STRING,
    source_column STRING,
    target_column STRING,
    target_data_type STRING,
    transform_expr STRING,
    is_business_key BOOLEAN,
    is_active BOOLEAN
)
```

```sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog_name || '.ict_audit.audit_pipeline_run')
(
    run_id LONG,
    domain STRING,
    pipeline_name STRING,
    job_run_id LONG,
    trigger_type STRING,
    triggered_by STRING,
    status STRING,
    start_ts TIMESTAMP,
    end_ts TIMESTAMP,
    message STRING
)
```

```sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog_name || '.ict_audit.audit_task_run')
(
    task_run_id LONG,
    run_id LONG,
    entity_id STRING,
    layer STRING,
    load_type STRING,
    watermark_from TIMESTAMP,
    watermark_to TIMESTAMP,
    rows_read STRING,
    rows_inserted STRING,
    rows_updated STRING,
    rows_deleted STRING,
    status STRING,
    start_ts TIMESTAMP,
    end_ts TIMESTAMP,
    message STRING
)
```

```sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog_name || '.ict_metadata.etl_watermark')
(
    entity_id STRING,
    layer STRING,
    watermark_column STRING,
    last_watermark_value TIMESTAMP,
    previous_watermark_value TIMESTAMP,
    last_run_id LONG,
    last_load_type STRING,
    last_success_ts TIMESTAMP,
    updated_by STRING
)
```
